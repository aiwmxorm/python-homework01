实验2：函数、条件、循环与递归
操作任务：
1. 编写score_level(score)函数：90〜100分返回“优秀”，80〜不⾜90分返回
“良好”，60〜不⾜80分返回“及格”，0〜不⾜60分返回“不及格”。对于超出0〜
100范围的数值，抛出ValueError。
2. 分别使⽤for和while计算1到100的整数和，⽐较两种实现的结果。
3. 编写递归版和迭代版阶乘函数，约定0! = 1，输⼊为⾮负整数。⽐较n = 0、1、
5、10时的结果。
4. 编写summary(values)函数，接收⾮空数值列表，以元组返回最⼩值、最⼤值和平
均值，并通过元组解包接收结果。
5. 编写sum_all(*args)函数，返回所有参数之和；分别测试⽆参数、单个参数和多
个参数的情况。

```python
# 1. 根据分数返回等级
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0 到 100 之间")

    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print("分数等级：")
for score in [95, 85, 70, 50]:
    print(f"{score} 分：{score_level(score)}")


# 2. 分别用 for 和 while 计算 1 到 100 的整数和
sum_for = 0
for number in range(1, 101):
    sum_for += number

sum_while = 0
number = 1
while number <= 100:
    sum_while += number
    number += 1

print("\n1 到 100 的整数和：")
print("for 循环：", sum_for)
print("while 循环：", sum_while)
print("结果相同：", sum_for == sum_while)


# 阶乘函数共用的输入检查
def check_factorial_input(n):
    if isinstance(n, bool) or not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")


# 3. 递归版阶乘
def factorial_recursive(n):
    check_factorial_input(n)
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


# 迭代版阶乘
def factorial_iterative(n):
    check_factorial_input(n)
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result


print("\n阶乘结果比较：")
for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)
    print(
        f"{n}!：递归版={recursive_result}，"
        f"迭代版={iterative_result}，"
        f"结果相同={recursive_result == iterative_result}"
    )


# 4. 返回最小值、最大值和平均值
def summary(values):
    if not values:
        raise ValueError("values 不能为空")

    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return minimum, maximum, average


minimum, maximum, average = summary([10, 20, 30, 40])
print("\n列表统计结果：")
print(f"最小值={minimum}，最大值={maximum}，平均值={average}")


# 5. 接收任意数量参数并求和
def sum_all(*args):
    return sum(args)


print("\n任意参数求和：")
print("无参数：", sum_all())
print("单个参数：", sum_all(5))
print("多个参数：", sum_all(1, 2, 3, 4))
```

实验3：列表、字典、元组与基础统计
使⽤以下模拟成绩数据，不需要输⼊真实学⽣信息：
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
操作任务：
1. 使⽤zip()构建姓名到成绩的字典，使⽤items()逐项输出姓名和成绩。
2. 使⽤enumerate()输出序号、姓名和成绩，序号从1开始。
3. 使⽤列表推导式提取及格学⽣的姓名，并⽣成所有成绩的平⽅列表。
4. 计算⼈数、平均分、最⾼分、最低分和及格率，使⽤字符串格式化将平均分显⽰为

```python
# 模拟成绩数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]


# 1. 用 zip() 构建姓名到成绩的字典，并用 items() 逐项输出
score_dict = dict(zip(names, scores))

print("1. 姓名和成绩：")
for name, score in score_dict.items():
    print(f"{name}: {score}")


# 2. 用 enumerate() 输出序号、姓名和成绩，序号从 1 开始
print("\n2. 学生序号、姓名和成绩：")
for number, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{number}. {name}: {score}")


# 3. 用列表推导式提取及格学生姓名，并生成成绩平方列表
passing_names = [
    name for name, score in zip(names, scores)
    if score >= 60
]
score_squares = [score ** 2 for score in scores]

print("\n3. 及格学生：", passing_names)
print("成绩平方列表：", score_squares)


# 4. 计算人数、平均分、最高分、最低分和及格率
student_count = len(scores)
average_score = sum(scores) / student_count
highest_score = max(scores)
lowest_score = min(scores)
passing_count = sum(1 for score in scores if score >= 60)
passing_rate = passing_count / student_count

print("\n4. 成绩统计：")
print(f"人数：{student_count}")
print(f"平均分：{average_score:.2f}")
print(f"最高分：{highest_score}")
print(f"最低分：{lowest_score}")
print(f"及格率：{passing_rate:.2%}")


# 5. 比较别名和副本：使用单独的演示列表
demo_scores = [85, 58, 92]
alias = demo_scores
copied = demo_scores.copy()

alias[0] = 100

print("\n5. 别名和副本：")
print("alias：", alias)
print("原列表 demo_scores：", demo_scores)
print("副本 copied：", copied)


# 6. 用元组保存学生记录，验证元组元素不能直接重新赋值
student_record = ("Alice", 85, "A班")

print("\n6. 学生记录元组：", student_record)

try:
    student_record[1] = 90
except TypeError as error:
    print("不能直接修改元组元素：", error)


# zip() 遇到长度不同的序列时，以较短序列为准
numbers = [1, 2, 3]
letters = ["a", "b"]

print("长度不同的序列使用 zip()：", list(zip(numbers, letters)))
```

```python
import os
import sys
import importlib
import pickle

# 1. 查看工作目录并创建数据目录
work_dir = os.getcwd()
print("当前工作目录：", work_dir)

data_dir = os.path.join(work_dir, "exp01_data")
os.makedirs(data_dir, exist_ok=True)

students_path = os.path.join(data_dir, "students.txt")
report_path = os.path.join(data_dir, "report.txt")
pickle_path = os.path.join(data_dir, "students.pkl")

# 2. 写入 UTF-8 学生成绩文件
student_data = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(students_path, "w", encoding="utf-8") as file:
    file.write(student_data)

print("\n已写入：", students_path)

# 3. 创建 student_utils.py 模块
module_path = os.path.join(work_dir, "student_utils.py")
module_source = '''def parse_student(line):
    """解析一行“姓名,成绩”，返回包含姓名和成绩的字典。"""
    fields = line.strip().split(",")

    if len(fields) != 2:
        raise ValueError("字段数量必须为 2")

    name = fields[0].strip()
    score_text = fields[1].strip()

    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(score_text)
    except ValueError:
        raise ValueError("成绩必须是整数")

    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")

    return {"name": name, "score": score}
'''

with open(module_path, "w", encoding="utf-8") as file:
    file.write(module_source)

# 导入刚刚创建的模块；重新运行单元格时也会加载最新内容
importlib.invalidate_caches()
if "student_utils" in sys.modules:
    importlib.reload(sys.modules["student_utils"])
else:
    import student_utils

from student_utils import parse_student

print("已创建模块：", module_path)

# 4. 逐行读取文件，跳过表头，记录有效和无效数据
valid_records = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as file:
    next(file)  # 跳过表头
    for line_number, line in enumerate(file, start=2):
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as error:
            invalid_records.append({
                "line": line_number,
                "content": line.rstrip("\n"),
                "error": str(error),
            })

print("\n有效记录：")
for record in valid_records:
    print(record)

print("\n无效记录及原因：")
for record in invalid_records:
    print(
        f"第 {record['line']} 行：{record['content']!r}，"
        f"错误：{record['error']}"
    )

# 5. 统计并写入报告
valid_count = len(valid_records)
invalid_count = len(invalid_records)
scores = [record["score"] for record in valid_records]

average_score = sum(scores) / valid_count if valid_count else 0
passing_count = sum(score >= 60 for score in scores)
passing_rate = passing_count / valid_count if valid_count else 0

report_content = (
    f"有效记录数：{valid_count}\n"
    f"无效记录数：{invalid_count}\n"
    f"平均分：{average_score:.2f}\n"
    f"及格率：{passing_rate:.2%}\n"
)

# w 模式：写入；如果文件已有内容，会先清空旧内容
with open(report_path, "w", encoding="utf-8") as file:
    file.write(report_content)

# a 模式：追加到文件末尾，不清除已有内容
with open(report_path, "a", encoding="utf-8") as file:
    file.write("模式说明：本报告先用 w 模式写入，再用 a 模式追加本行。\n")

print("\n报告内容：")
with open(report_path, "r", encoding="utf-8") as file:
    print(file.read())

# 6. 用 pickle 保存并读取有效记录
with open(pickle_path, "wb") as file:
    pickle.dump(valid_records, file)

with open(pickle_path, "rb") as file:
    restored_records = pickle.load(file)

print("pickle 恢复的数据与原数据相等：", restored_records == valid_records)

# 7. 测试从 student_utils 模块导入的函数
print("\n模块函数测试：")
print("有效输入：", parse_student("Zoe,88"))

try:
    parse_student("Noah,abc")
except ValueError as error:
    print("无效输入：", error)

# 8. 尝试读取一个明确用于演示的文件，并捕获文件不存在异常
missing_path = os.path.join(data_dir, "__missing_file_for_exp01__.txt")

try:
    with open(missing_path, "r", encoding="utf-8") as file:
        file.read()
except FileNotFoundError:
    print(f"\n文件不存在（符合预期）：{missing_path}")
```

实验5：类、继承与对象复制
操作任务：
1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
和__str__()⽅法。
2. 为实验4中的每条有效记录创建⼀个Student对象，输出对象信息和及格状态。
3. 定义继承⾃Student的GraduateStudent类，增加research_topic属性，通过
super().__init__()初始化⽗类属性，并重写__str__()。
4. 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分
别进⾏直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，⽐较结果。
每次⽐较都从新建的原始对象开始。

```python
import copy
import os
import pickle


# 1. 定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"学生：{self.name}，成绩：{self.score}"


# 2. 从实验4的 pickle 文件读取有效记录并创建 Student 对象
pickle_path = os.path.join(
    os.getcwd(), "exp01_data", "students.pkl"
)

with open(pickle_path, "rb") as file:
    valid_records = pickle.load(file)

students = [
    Student(record["name"], record["score"])
    for record in valid_records
]

print("2. 实验4有效记录对应的学生对象：")
for student in students:
    print(f"{student}，及格：{student.is_passed()}")


# 3. 定义继承自 Student 的 GraduateStudent 类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"研究生：{self.name}，成绩：{self.score}，"
            f"研究方向：{self.research_topic}"
        )


graduate = GraduateStudent("Alice", 95, "数据分析")
print("\n3. 研究生对象：")
print(f"{graduate}，及格：{graduate.is_passed()}")


# 4. 比较直接赋值、浅拷贝和深拷贝
print("\n4. 对象复制比较：")

# 直接赋值：两个变量指向同一个对象
original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"].append(95)

print("\n直接赋值后：")
print("original：", original)
print("assigned：", assigned)


# 浅拷贝：外层字典不同，但内部列表仍是同一个列表
original = {"name": "Alice", "scores": [85, 90]}
shallow_copied = copy.copy(original)
shallow_copied["scores"].append(95)

print("\ncopy.copy() 后：")
print("original：", original)
print("shallow_copied：", shallow_copied)
print("外层对象相同：", original is shallow_copied)
print("内部列表相同：", original["scores"] is shallow_copied["scores"])


# 深拷贝：外层字典和内部列表都复制
original = {"name": "Alice", "scores": [85, 90]}
deep_copied = copy.deepcopy(original)
deep_copied["scores"].append(95)

print("\ncopy.deepcopy() 后：")
print("original：", original)
print("deep_copied：", deep_copied)
print("外层对象相同：", original is deep_copied)
print("内部列表相同：", original["scores"] is deep_copied["scores"])
```

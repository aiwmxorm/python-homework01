热⾝：定义简单函数
1. 定义⼀个名为 say_hi 的函数，它不接受任何参数，并在被调⽤时打印字符串
"Hello, world!"。
2. 定义⼀个名为 goat_pad 的函数，它接受⼀个字符串作为其唯⼀参数，并打印该字
符串，前后都加上字符串 "goat"。
3. 定义⼀个名为 print_n 的函数，它接受两个参数，⼀个字符串 s 和⼀个整数 n，并
打印字符串 n 次，每次在单独的⼀⾏上。

```python
# 1. 不接受参数，打印问候语
def say_hi():
    print("Hello, world!")


# 2. 接受一个字符串，前后加上 "goat" 后打印
def goat_pad(text):
    print(f"goat{text}goat")


# 3. 将字符串重复打印 n 次，每次一行
def print_n(s, n):
    for _ in range(n):
        print(s)


# 调用函数查看结果
say_hi()
goat_pad("hello")
print_n("Python", 3)
```

欧⼏⾥得算法
欧⼏⾥得算法是⼀种寻找两个数的最⼤公约数（GCD）的⽅法。回想⼀下，两个数m和n
的最⼤公约数是同时能整除m和n的最⼤数。
实现⼀个名为gcd的函数，它接受两个整数作为其参数并返回它们的GCD。你可以
假设两个输⼊都是整数，所以你不需要在你的函数中包含任何错误检查。


```python
def gcd(m, n):
    """使用欧几里得算法返回 m 和 n 的最大公约数。"""
    m = abs(m)
    n = abs(n)

    while n != 0:
        m, n = n, m % n

    return m


# 测试指定的数字对
pairs = [(2018, 2019), (1200, 300), (5040, 60)]

for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")


# 负数输入测试
print("gcd(-18, 12) =", gcd(-18, 12))
print("gcd(-18, -12) =", gcd(-18, -12))
```

近似欧拉数e

```python
import math


# 1. 使用 (1 + 1/n) ** n 近似 e
# n 应为正整数
def euler_limit(n):
    return (1 + 1 / n) ** n


# 2. 使用公式（1）的前 n 项近似 e
# n=4 时，计算 1/0! + 1/1! + 1/2! + 1/3!
def euler_infinite_sum(n):
    total = 0.0
    term = 1.0  # 当前项为 1/0!

    for k in range(n):
        total += term
        term /= k + 1  # 更新为下一项 1/(k+1)!

    return total


# 3. 累加公式（1），直到近似值与真实 e 的误差小于 epsilon
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    term = 1.0  # 1/0!
    k = 0

    while abs(total - true_e) >= epsilon:
        total += term
        k += 1
        term /= k

    return total


# 4. 打印前 1 到 n 项的近似值，每行一个
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"前 {k} 项：{euler_infinite_sum(k):.10f}")


# 测试
print("1. 极限公式：")
print(f"euler_limit(1000) = {euler_limit(1000):.10f}")

print("\n2. 无穷级数的前 n 项：")
for n in [4, 1, 0]:
    print(f"euler_infinite_sum({n}) = {euler_infinite_sum(n):.10f}")

print("\n3. 误差小于 1e-6 的近似值：")
epsilon = 1e-6
approximation = euler_approx(epsilon)
print(f"近似值：{approximation:.10f}")
print(f"真实值：{math.exp(1):.10f}")
print(f"绝对误差：{abs(approximation - math.exp(1)):.10g}")

print("\n4. 前 8 项近似值表：")
print_euler_sum_table(8)
```

测试整数属性

```python
def is_power_of_2(n):
    # 1 = 2**0，也是 2 的幂
    while n > 1 and n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    # 1 的任意正整数次幂都是 1
    if b == 1:
        return n == 1

    while n > 1 and n % b == 0:
        n //= b
    return n == 1


# 测试 is_power_of_2
for number in [1, 2, 8, 16, 18]:
    print(f"is_power_of_2({number}) = {is_power_of_2(number)}")

print()

# 测试 is_power(b, n)：判断 n 是否为 b 的幂
for b, n in [(2, 16), (3, 27), (5, 25), (2, 12), (1, 1), (1, 8)]:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```

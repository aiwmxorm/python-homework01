字符串的乐趣。
Python 切片一般写作 字符串[开始:结束:步长]。两个冒号分别分隔这三个部分。在 processed[::-1] 中，开始和结束都省略了，步长是 -1，表示从后往前取，所以会反转字符串。replace(要替换的内容, 替换成什么) 是字符串方法。replace 会返回一个新字符串，不会直接修改原来的 s。

```python
# 1. 判断回文：忽略空格、忽略大小写
def is_palindrome(s):
    # 去掉所有空格，全部转为小写
    processed = s.replace(" ", "").lower()
    # 和反转后的字符串对比
    return processed == processed[::-1]


# 2. 判断是否字母顺序（abecedarian）：忽略空格，不区分大小写
def is_abecedarian(s):
    # 去掉空格，转小写
    processed = s.replace(" ", "").lower()
    # 依次比较每个字符是否 >= 前一个
    for i in range(1, len(processed)):
        if processed[i] < processed[i-1]:
            return False
    return True


# 3. 删除元音 a e i o u（大小写元音都删掉）
def remove_vowels(s):
    vowels = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
    res = ""
    for char in s:
        if char not in vowels:
            res += char
    return res


# ========== 测试样例 ==========
if __name__ == "__main__":
    print(is_palindrome("T A C O cat"))   # True
    print(is_palindrome("level"))         # True
    print(is_palindrome("hello"))         # False

    print(is_abecedarian("abcd efgh xyz"))# True
    print(is_abecedarian("adder"))        # True
    print(is_abecedarian("dog"))          # False

    print(remove_vowels("cat"))           # 'ct'
    print(remove_vowels("audio"))         # 'd'
    print(remove_vowels("AEIOUaeiou"))    # ''
```

列表的乐趣。isinstance() 是 Python 内置函数，用来判断一个值是否属于指定类型。isinstance(要检查的值, 类型)。

```python
# 1. list_reverse：接收列表，返回反转列表；非列表抛出错误
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]


# 2. is_sorted：只遍历1次，判断序列是否升序
def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i] < seq[i-1]:
            return False
    return True


# ===== 测试代码 =====
if __name__ == "__main__":
    print(list_reverse([1,2,3]))    # [3, 2, 1]
    # list_reverse(123)  # 会触发 TypeError

    print(is_sorted([1,2,3,4]))     # True
    print(is_sorted([1,3,2]))       # False
    print(is_sorted(("a", "b", "c"))) # True，元组也可以（序列）
    print(is_sorted([5]))           # True，单个元素默认升序
```

作为向量的元组。tuple(...)：把生成的结果收集成元组。

```python
# 1. 向量标量乘法 vec_scalar_mult
def vec_scalar_mult(t, s):
    # 检查 t 是否为元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查 s 是否为 int/float
    if not isinstance(s, (int, float)):
        raise TypeError("标量 s 必须是整数或浮点数")
    # 检查元组内每个元素是 int/float
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组中的元素必须是整数或浮点数")
    # 逐元素相乘，返回新元组
    return tuple(elem * s for elem in t)


# 2. 向量内积 vec_inner_product
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度不一致")
    total = 0.0
    for xi, yi in zip(x, y):
        # 校验元素类型
        if not (isinstance(xi, (int, float)) and isinstance(yi, (int, float))):
            raise TypeError("向量元素必须是整数或浮点数")
        total += xi * yi
    return total


# 3. 检查有效矩阵 check_valid_mx
def check_valid_mx(mx):
    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 空矩阵认为合法
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        # 每一行必须是元组
        if not isinstance(row, tuple):
            return False
        # 每行长度要一致
        if len(row) != row_len:
            return False
        # 每个元素是 int/float
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


# 4. 矩阵乘向量 mx_vec_mult
def mx_vec_mult(mx, v):
    # 先校验矩阵合法性
    if not check_valid_mx(mx):
        raise TypeError("不是一个合法矩阵")
    # 校验向量是元组
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")
    # 矩阵行数可以任意，矩阵列数 = 向量长度
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("矩阵的列数必须等于向量长度")
    # 矩阵 m×n，向量 n×1，结果 m×1：每行和向量做内积
    res = []
    for row in mx:
        res.append(vec_inner_product(row, v))
    return tuple(res)


# ========= 测试示例 =========
if __name__ == "__main__":
    print(vec_scalar_mult((1, 2, 3), 2))          # (2, 4, 6)
    print(vec_inner_product((1, 2), (3, 4)))      # 11.0
    print(check_valid_mx(((1.0,2),(3,4))))        # True
    print(mx_vec_mult(((1,2),(3,4)), (5,6)))      # (17.0, 39.0)
```

更多向量的乐趣。通常应把维度和元素作为实例属性，因为不同矩阵对象可以有不同的行数、列数和元素值。每个实例都应保存自己的数据，互不影响。

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 检查维度是否为非负整数
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")

        # 默认元素：全0
        if elements is None:
            elements = [0.0 for _ in range(dim)]
        else:
            # 转为列表，校验每个元素是int/float
            elements = list(elements)
            for num in elements:
                if not isinstance(num, (int, float)):
                    raise TypeError("向量元素只能是整数或浮点数")

        # 校验元素数量和维度匹配
        if len(elements) != dim:
            raise ValueError("元素个数必须和维度相等")

        # 实例属性：每个向量自己的维度和元素
        self.dim = dim
        self.elements = elements

    # 2. 维度dim、元素elements是【实例属性】
    # 理由：不同Vector实例可以拥有不同维度、不同元素值；
    # 如果是类属性，所有向量共享同一套dim和elements，无法独立表示不同向量。

    # 3. 比较运算符：__eq__, __lt__, __le__, __gt__, __ge__
    def _check_same_dim(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不同，无法比较")

    def __eq__(self, other):
        try:
            self._check_same_dim(other)
        except (TypeError, ValueError):
            return False
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dim(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_same_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_same_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_same_dim(other)
        return self.elements >= other.elements

    # 4. dot 内积方法
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot的参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("内积要求两个向量维度相同")
        total = 0.0
        for a, b in zip(self.elements, other.elements):
            total += a * b
        return total

    # 5. 乘法：*
    # v * scalar 或者 v * w（向量逐元素乘）
    def __mul__(self, other):
        if isinstance(other, (int, float)):
            # 标量右乘 v * 2
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            # 向量 * 向量：逐元素相乘
            if self.dim != other.dim:
                raise ValueError("逐元素乘法要求向量维度相同")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            return NotImplemented

    # 标量左乘：2 * v，调用 __rmul__
    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return self * other
        return NotImplemented

    # 方便打印查看
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"


# ========= 测试代码 =========
if __name__ == "__main__":
    v1 = Vector(2, [2, 4])
    v2 = Vector(2, [2, 5])
    print(v1 < v2)          # True
    print(v1 == v2)         # False

    v3 = Vector(2, [2,3])
    v4 = Vector(2, [4,5])
    print(v3 * v4)          # Vector(2, [8, 15])
    print(v3 * 2)           # Vector(2, [4, 6])
    print(2 * v3)           # Vector(2, [4, 6])

    print(v3.dot(v4))       # 2*4 + 3*5 = 23.0

    v_zero = Vector(3)      # 不传元素，默认 [0.0,0.0,0.0]
    print(v_zero)
```
